```python
import joblib
```

```python
final_svm = joblib.load("final_phishing_svm_model.pkl")
tfidf_final = joblib.load("final_tfidf_vectorizer.pkl")
```

```python
def predict_url(url):
    url = url.lower()
    url_vec = tfidf_final.transform([url])
    prediction = final_svm.predict(url_vec)
    return prediction[0]


predict_url("www.youtube.com/")
```
